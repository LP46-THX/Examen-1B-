```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

uploaded = files.upload()

df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

df.head()

df.info()

df.describe()





```

```python

total_empleados = len(df)
empleados_attrition = len(df[df['Attrition'] == 'Yes'])
tasa_rotacion = empleados_attrition / total_empleados * 100



tabla_rotacion = pd.DataFrame({
    'Total Empleados':[total_empleados],
    'Empleados que salieron':[empleados_attrition],
    'Tasa de Rotación (%)':[round(tasa_rotacion,2)]
})
tabla_rotacion
```

Analiza la cantidad de empleados que han desertado de la empresa mediante la formula de tasa de rotacion de empleados, en este caso la empresa sufre una reduccion del 16.12% del personal, esta en un rango moderado de rotacion de personal


```python
tabla_ingresos = df.groupby('Department')['MonthlyIncome'].mean().reset_index()
tabla_ingresos.rename(columns={'MonthlyIncome':'Ingreso Promedio Mensual'}, inplace=True)
print(tabla_ingresos)


plt.bar(tabla_ingresos['Department'], tabla_ingresos['Ingreso Promedio Mensual'], color='blue')
plt.ylabel('Ingreso Mensual Promedio')
plt.title('KPI 2 - Ingreso Promedio por Departamento')
plt.xticks(rotation=45)
plt.show()
```
